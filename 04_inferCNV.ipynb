```python
import scanpy as sc
```

```python
adata = sc.read_h5ad("/home/matei/data/adata_QC_filtered.h5ad")
```

```python
adata.X.max()
```

```python
# Calculating lncRNA ratio
lncRNA = adata.var[adata.var["transcript_type"]=="lncRNA"].index
total = adata.var.index
lncRNA_exp = adata[:, lncRNA].X.sum(axis=1)
total_exp = adata[:, total].X.sum(axis=1)
ratio = lncRNA_exp/total_exp
```

```python
adata.obs["ncRNA_exp"] = ratio
```

```python
adata = adata[adata.obs["ncRNA_exp"]<0.35]
```

```python
import pandas as pd

counts = pd.DataFrame.sparse.from_spmatrix(
    adata.X,
    index=adata.obs_names,
    columns=adata.var["gene"]
)

counts_gene = counts.groupby(level=0, axis=1).sum()
```

```python
counts_gene.T.to_csv(
    "infercnv_counts.tsv",
    sep="\t"
)
```
